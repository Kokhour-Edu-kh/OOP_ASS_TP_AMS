$$TP01.3. Asterisks$$
Write a program in Java to display shapes as below:

```python
# A .

for i in range(1,7):
    for j in range(1,21):
        if i==1 or i==6 or j==1 or j==11:
            print("*",end="")
        else:
            print(" ",end=" ")
    print()
```

```python
# B .


for i in range(1,6):
    #space
    for j in range(5-i):
        print(" ",end=" ")
    #star

    for j in range(2*i-1):
        print("*",end=" ")

    print()
```

```python
# C

for i in range(1,6):
    print("*" * (2 * i -1))
```

```python
# D


for i in range(9, -1, -2):
    for j in range(0, 10):
        if j >= i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# E
# Upper half
for i in range(1, 4):

    for j in range(2 * (3 - i)):
        print(" ", end="")

    for j in range(2 * i - 1):
        print("*", end="")

    print()

# Lower half
for i in range(2, 0, -1):
    for j in range(2 * (3-i)):
        print(" ", end="")
    for j in range(2 * i-1):
        print("*", end="")

    print()
```

```python
# F

# Upper half
for i in range(1,4):
    for j in range(3-i):
        print(" ", end="")

    for j in range(2 *i-1):
        print("*", end="")
    print()


# Lower half
for i in range(2, 0, -1):
    for j in range(3-i):
        print(" ", end="")

    for j in range(2 * i - 1):
        print("*", end="")
    print()
```

```python
#G

# Upper inverted triangle
for i in range(3,0,-1):

    for j in range(3- i):
        print(" ", end="")

    for j in range(2*i - 1):
        print("*", end="")

    print()


# Lower upright triangle
for i in range(2, 4):

    for j in range(3- i):
        print(" ", end="")

    for j in range(2*i-1):
        print("*", end="")

    print()
```
