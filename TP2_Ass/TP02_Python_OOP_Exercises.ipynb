## Exercise 1: Greeting a User
### Objective & Concept
- **Goal**: Read a user's name, store it in an object attribute, and display `Hello, <name>!`.
- **Key Concept**: In Python OOP, `__init__` is the constructor method called when an object is instantiated. `self.name = name` assigns the passed argument to an instance attribute.

```python

class Greeting:
    """A class for greeting a user."""

    def __init__(self, name):
        self.name = name

    def say_hello(self):
        """Print a greeting message."""
        print(f"Hello, {self.name}!")

def main():
    """Get the user's name and display a greeting."""
    name = input("Enter your name: ").strip()

    greeting = Greeting(name)
    greeting.say_hello()


if __name__ == "__main__":
    main()
```

## Exercise 2: Rectangle
### Objective & Concept
- **Goal**: Store `width` and `height`, then compute perimeter $P = 2(\text{width} + \text{height})$ and area $A = \text{width} \times \text{height}$.
- **Key Concept**: Methods can calculate and return values (`calculate_perimeter`, `calculate_area`), while another method (`display_results`) formats and prints them.

```python
class Rectangle:
    """A rectangle with a length and a width."""

    def __init__(self, length, width):
        if length <= 0 or width <= 0:
            raise ValueError("length and width must be greater than 0")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def display(self):
        print(f"Length    : {self.length}")
        print(f"Width     : {self.width}")
        print(f"Area      : {self.area()}")
        print(f"Perimeter : {self.perimeter()}")


def read_number(message):
    """Ask until the user types a valid number."""
    while True:
        try:
            return float(input(message))
        except ValueError:
            print("Please enter a number, for example 5 or 3.5")


def main():
    length = read_number("Enter the length of the rectangle: ")
    width = read_number("Enter the width of the rectangle: ")
    try:
        rect = Rectangle(length, width)
    except ValueError as error:
        print("Error:", error)
    else:
        rect.display()


if __name__ == "__main__":
    main()
```

## Exercise 3: Equation Calculator
### Objective & Concept
- **Equation**: $\frac{1}{x} = \frac{1}{y} + \frac{1}{z} \implies x = \frac{yz}{y+z}$.
- **Validation Rules**: Division by zero must be prevented:
  1. $y \neq 0$
  2. $z \neq 0$
  3. $y + z \neq 0$

```python
class EquationCalculator:
    def __init__(self, y: float, z: float):
        """Constructor initializing y and z."""
        self.y = y
        self.z = z

    def calculate_x(self):
        """Calculates x while guarding against division by zero."""
        if self.y == 0 or self.z == 0 or (self.y + self.z) == 0:
            return None
        return (self.y * self.z) / (self.y + self.z)

    def display_result(self):
        
        """Displays the calculated value or validation error."""
        x = self.calculate_x()
        if x is None:
            print("Error: Invalid values! y and z must be non-zero and (y + z) != 0.")
        else:
            print(f"x = {x:.4g}")


# --- Test Cell for Exercise 3 ---
try:
    val_y = float(input("Enter y: "))
    val_z = float(input("Enter z: "))
    calc = EquationCalculator(val_y, val_z)
    calc.display_result()
    
except ValueError:
    print("Error: Please enter valid numbers.")
```

## Exercise 4: Hundreds Counter
### Objective & Concept
- **Goal**: Count how many complete hundreds exist in a non-negative integer (e.g., $785 \to 7$).
- **Key Concept**: Use floor division `// 100` and validate that `number >= 0`.

```python
class HundredsCounter:
    """Counts how many complete hundreds are in a non-negative number."""
    def __init__(self, number):
        if number < 0:
            raise ValueError("Number must be non-negative.")

        self.number = number

    def count_hundreds(self):
        return self.number // 100


"""Ask the user for a number and print its complete hundreds."""
try:
    number=int(input("Enter a non-negative Number : "))
    counter = HundredsCounter(number)
    print("Complete Hundreds in the number:", counter.count_hundreds())
except ValueError as e:
    print("Error : ", e)
```

## Exercise 5: Cheater Game (Lucky Number)
### Objective & Concept
- **Goal**: Read an integer and display `number + 1`.
- **Key Concept**: Encapsulate the calculation in `get_next_number()` and output via `display_result()`.

```python
class LuckyNumberGame:
    def __init__(self, number: int):
        """Constructor initializing the input number."""
        self.number = number

    def get_next_number(self) -> int:
        """Returns the number plus one."""
        return self.number + 1

    def display_result(self):
        """Displays the next number."""
        print(f"The next number is: {self.get_next_number()}")


# --- Test Cell for Exercise 5 ---
try:
    user_int = int(input("Enter an integer: "))
    game = LuckyNumberGame(user_int)
    game.display_result()
except ValueError:
    print("Error: Please enter a valid integer.")
```
